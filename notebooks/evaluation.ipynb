## Этапы и что было сделано
1. Zero-shot классификация.

5. Квантование и LoRA (скорость инференса + время обучения).


```python
from pathlib import Path
import pandas as pd
import matplotlib.pyplot as plt

# Укажите путь к папке запуска
RUN_DIR = Path("../outputs/PUT_RUN_DIR_HERE")

zero_shot = pd.read_csv(RUN_DIR / "zero_shot.csv") if (RUN_DIR / "zero_shot.csv").exists() else None
freeze_layers = pd.read_csv(RUN_DIR / "freeze_layers.csv") if (RUN_DIR / "freeze_layers.csv").exists() else None
adapters = pd.read_csv(RUN_DIR / "adapters.csv") if (RUN_DIR / "adapters.csv").exists() else None
datasets = pd.read_csv(RUN_DIR / "datasets.csv") if (RUN_DIR / "datasets.csv").exists() else None
quant_lora = pd.read_csv(RUN_DIR / "quant_lora.csv") if (RUN_DIR / "quant_lora.csv").exists() else None
summary = pd.read_csv(RUN_DIR / "summary.csv") if (RUN_DIR / "summary.csv").exists() else None
```

## Заморозка слоев: метрики и время


```python
if freeze_layers is not None:
    for fmt in sorted(freeze_layers["data_format"].unique()):
        subset = freeze_layers[freeze_layers["data_format"] == fmt]
        plt.plot(subset["frozen_layers_pct"], subset["accuracy"], marker="o", label=f"acc {fmt}")
    plt.title("Accuracy vs Frozen Layers")
    plt.xlabel("Frozen layers (%)")
    plt.ylabel("Accuracy")
    plt.legend()
    plt.show()

    for fmt in sorted(freeze_layers["data_format"].unique()):
        subset = freeze_layers[freeze_layers["data_format"] == fmt]
        plt.plot(subset["frozen_layers_pct"], subset["training_time_s"], marker="o", label=f"time {fmt}")
    plt.title("Training time vs Frozen Layers")
    plt.xlabel("Frozen layers (%)")
    plt.ylabel("Time (s)")
    plt.legend()
    plt.show()
```

## Адаптеры: метрики и время


```python
if adapters is not None:
    for adapter in adapters["approach"].unique():
        subset = adapters[adapters["approach"] == adapter]
        plt.plot(range(len(subset)), subset["accuracy"], marker="o", label=adapter)
    plt.title("Adapters Accuracy")
    plt.ylabel("Accuracy")
    plt.legend()
    plt.show()

    for adapter in adapters["approach"].unique():
        subset = adapters[adapters["approach"] == adapter]
        plt.plot(range(len(subset)), subset["training_time_s"], marker="o", label=adapter)
    plt.title("Adapters Training Time")
    plt.ylabel("Time (s)")
    plt.legend()
    plt.show()
```

## Сравнение датасетов


```python
if datasets is not None:
    plt.bar(datasets["approach"], datasets["accuracy"])
    plt.title("Accuracy by Dataset")
    plt.show()

    plt.bar(datasets["approach"], datasets["f1"])
    plt.title("F1 by Dataset")
    plt.show()

    plt.bar(datasets["approach"], datasets["training_time_s"])
    plt.title("Training Time by Dataset")
    plt.show()
```
